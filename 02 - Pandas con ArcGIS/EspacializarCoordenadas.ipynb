```python
import pandas as pd
import os
```

```python
df = pd.read_json("https://www.datos.gov.co/resource/pi36-fdpk.json")
df
```

```python
df["X_Lon"] = df["punto"].str["longitude"]
df["Y_Lat"] = df["punto"].str["latitude"]

df["X_Lon"] = pd.to_numeric(df["X_Lon"])
df["Y_Lat"] = pd.to_numeric(df["Y_Lat"])

df
```

```python
csv_temporal = os.path.join(arcpy.env.scratchFolder, "datos.csv")
df.to_csv(csv_temporal)
```

```python
arcpy.management.XYTableToPoint(
    in_table = csv_temporal, 
    out_feature_class = "puntos", 
    x_field = "X_Lon", 
    y_field = "Y_Lat"
)
```
