```python
import requests
import pandas as pd
import os
```

```python
url = "https://earthquake.usgs.gov/earthquakes/feed/v1.0/summary/significant_month.geojson"
respuesta = requests.get(url)
terremotos = respuesta.json()

terremotos
```

```python
lista_terremotos = []
```

```python
for terremoto in terremotos["features"]:
    coordenadas = terremoto["geometry"]["coordinates"]
    lugar = terremoto["properties"]["place"]
    magnitude = terremoto["properties"]["mag"]

    fila = {
        "magnitude": magnitude,
        "lugar": lugar,
        "X_Lon": coordenadas[0],
        "Y_Lat": coordenadas[1]
    }
    
    lista_terremotos.append(fila)
```

```python
df = pd.DataFrame(lista_terremotos)
df
```

```python
csv_temporal = os.path.join(arcpy.env.scratchFolder, "datos.csv")
df.to_csv(csv_temporal)
```

```python
arcpy.management.XYTableToPoint(
    in_table = csv_temporal, 
    out_feature_class = "puntos", 
    x_field = "X_Lon", 
    y_field = "Y_Lat"
)
```
